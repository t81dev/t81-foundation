## 5. Agent Structure

An agent is a persistent process with state (memory) and a policy (Axion).

```python
# Conceptual Agent Loop
class Agent:
    def __init__(self, seed):
        self.seed = seed
        self.state = 0
    
    def step(self, input_val):
        # Deterministic update based on seed + input
        self.state = (self.state + input_val + self.seed) % 100
        return self.state

a = Agent(seed=42)
print(f"Step 1: {a.step(10)}")
print(f"Step 2: {a.step(20)}")
```

## 6. Entropy Accounting

If an agent needs "fresh" randomness, it consumes entropy from the block header, which is deterministic for all validators.

## 7. Policy Governance

Agents are not free to do anything. Axion policies wrap every action. An agent trying to spend more than its budget is blocked.
