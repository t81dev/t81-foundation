## 5. Exploratory Code: BigInt Operations

`BigInt` supports arbitrary-precision integers.

```python
x = BigInt(1234567890123456789)
y = BigInt(9876543210987654321)

prod = x * y
print(f"Product: {prod}")

diff = x - y
print(f"Difference: {diff}")
```

## 6. Exploratory Code: Fraction Reduction

`Fraction` (T81Frac81) automatically reduces fractions to their simplest form.

```python
# Construct a fraction 50/100, which should reduce to 1/2
f1 = Fraction(50, 100)
print(f"Fraction(50, 100) -> {f1}")

# Arithmetic with fractions
f2 = Fraction(1, 3)
sum_frac = f1 + f2  # 1/2 + 1/3 = 5/6
print(f"{f1} + {f2} = {sum_frac}")

prod_frac = f1 * f2 # 1/2 * 1/3 = 1/6
print(f"{f1} * {f2} = {prod_frac}")
```

## 7. Trace or Hash Validation

Checking ring properties for consistency.

```python
# Associativity: (a + b) + c == a + (b + c)
a = Fraction(1, 7)
b = Fraction(2, 5)
c = Fraction(3, 8)

left = (a + b) + c
right = a + (b + c)

print(f"Left: {left}")
print(f"Right: {right}")
# Since we don't have direct equality operator exposed in python yet, compare strings
assert str(left) == str(right)
print("Associativity Verified")
```
