## 7. Memory Quota Simulations

T81 containers track their memory usage. Exceeding a quota triggers a trap.

```python
# Simulation
class QuotaList(list):
    def __init__(self, max_items):
        self.max_items = max_items
    
    def append(self, item):
        if len(self) >= self.max_items:
            raise MemoryError("Quota Exceeded")
        super().append(item)

qlist = QuotaList(2)
try:
    qlist.append(1)
    qlist.append(2)
    qlist.append(3)
except MemoryError as e:
    print(f"Caught: {e}")
```

## 8. Architectural Commentary

Deterministic containers are essential for consensus. If two nodes iterate a map in different orders, their computed hashes will diverge, causing a fork. T81 enforces sorted iteration at the container level.
