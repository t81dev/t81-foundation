## 5. Hash Derivation

T81 uses a specific hashing algorithm (e.g., SHA3-256 or BLAKE3) to derive content addresses.

```python
def calculate_canon_hash(data: bytes) -> str:
    # Simulating CanonHash
    return hashlib.sha256(data).hexdigest()

content = b"Hello T81 CanonFS"
addr = calculate_canon_hash(content)
print(f"Content: {content}")
print(f"Address: {addr}")
```

## 6. Snapshot Lineage

A snapshot is a Merkle root of the filesystem state. Changing one file changes the root hash.

```python
root_v1 = calculate_canon_hash(b"file1_v1")
print(f"Root V1: {root_v1}")

root_v2 = calculate_canon_hash(b"file1_v2")
print(f"Root V2: {root_v2}")

assert root_v1 != root_v2
print("Root hash changed with content (Correct).")
```

## 7. Deterministic Storage

Retrieving by hash guarantees we get the exact bytes we asked for.

## 8. Architectural Commentary

CanonFS abstracts away the physical location of data. Whether on local disk, IPFS, or S3, the data is identified only by its hash. This is key for decentralized model distribution.
